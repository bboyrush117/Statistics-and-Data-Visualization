# Independent Samples t-Test Practice

A fitness coach wants to determine whether a new workout program leads to greater weight loss compared to a standard workout program.

Participants were randomly assigned to one of two workout programs, and the amount of weight lost (in pounds) after 8 weeks is recorded below.

## Standard Workout Program
4, 5, 6, 5, 4, 7, 5, 6

## New Workout Program
8, 9, 7, 10, 8, 9, 11, 8

## Assumption
Assume the population variances are equal.

## Task

Conduct an independent two-sample t-test at the 0.05 significance level to determine whether there is a significant difference in mean weight loss between the two workout programs.

```python
import math

def mean(data):
  mean = sum(data)/len(data)
  return mean

def variance(data):
  m = mean(data)
  variance = sum((x-m)**2 for x in data)/(len(data) - 1)
  return variance

def standard_deviation(data,data_2):
  n_1 = len(data)
  n_2 = len(data_2)
  s_1 = variance(data)
  s_2 = variance(data_2)
  pooled_standard_deviation = math.sqrt(((n_1-1)*s_1 + (n_2-1)*s_2)/(n_1 + n_2 -2))
  return pooled_standard_deviation

def t_test_statistic(data,data_2):
  n_1 = len(data)
  n_2 = len(data_2)
  mu_1 = mean(data)
  mu_2 = mean(data_2)
  pooled_standard_deviation = standard_deviation(data,data_2)
  t_test_statistic = (mu_1 - mu_2)/(pooled_standard_deviation * math.sqrt(1/n_1 + 1/n_2))
  degrees_of_freedom = n_1 + n_2 - 2
  return t_test_statistic, degrees_of_freedom

standard = [4, 5, 6, 5, 4, 7, 5, 6]
new = [8, 9, 7, 10, 8, 9, 11, 8]

t_test_statistic, degrees_of_freedom = t_test_statistic(standard, new)

print("t-statistic:", t_test_statistic)
print("degrees of freedom:", degrees_of_freedom )
```

```python
from scipy.stats import ttest_ind

standard = [4, 5, 6, 5, 4, 7, 5, 6]
new = [8, 9, 7, 10, 8, 9, 11, 8]

t_stat, p_value = ttest_ind(standard, new, equal_var = True)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```
